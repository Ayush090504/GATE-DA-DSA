```python
arr = [12, 5, 8, 20, 3, 15]
max=0
for i in arr:
    if i>max:
        max=i
print(f"Maximum value in array: {max}")
```

```python
arr = [10, 25, 7, 18, 30, 42]
target=int(input())
index=0
for i in range(len(arr)):
    if arr[i]==target:
        index=i
if index==0:
    print("Not Found")
else:
    print(f"Found at index {index}")
```

```python
arr = [10, 20, 30, 40, 50]
target_index=int(input())
for i in range (len(arr)):
    if i == target_index:
        arr+=[None]
        for j in range(len(arr)-1,target_index,-1):
            arr[j]=arr[j-1]
        arr[target_index]=int(input())
print(arr)
```

```python
arr = [10, 20, 30, 40, 50]
target_index=int(input())
for i in range(len(arr)):
    if i ==target_index:
        for j in range(target_index,len(arr)-1):
            arr[j]=arr[j+1]
arr=arr[:-1]
print(arr)
```
