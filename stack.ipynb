```python
def push(val, s):
    top=len(s)-1
    s+=[None]
    top+=1
    s[top]=val

def pop(s):
    top=len(s)-1
    elem=s[top]
    s[:]=s[:-1]
    return elem

def peek(s):
    top=len(s)-1
    elem=s[top]
    return elem

s1=[1,-3,5]
s2=[2,4,6]

i=peek(s1)+pop(s2)
push(pop(s2),s1)
push(i,s2)
push(peek(s1),s2)
pop(s1)
j=peek(s2)+pop(s1)
push(peek(s2),s1)
a=s1[len(s1)-1]
b=s2[len(s2)-1]
print(s1)
print(s2)
print(i+j+a+b)
```

```python
max_size=int(input("Enter the maximum size of stack: "))
top=-1
stack=[0]*max_size
def push(val, stack, top):
    top+=1
    stack[top]=val
    return top
def pop(stack,top):
    val=stack[top]
    stack[top]=None
    top-=1
    return val,top
def peek(stack,top):
    val=stack[top]
    return val
while(1):
    op=int(input("Enter the operation (1->Push, 2->Pop, 3->Peek, 4->Exit): "))
    if op==1:
        if top>=max_size-1:
            print("Stack Overflow!!")
        else:
            val=int(input("Enter the value: "))
            top=push(val,stack,top)
    elif op==2:
        if top<0:
            print("Stack Underflow!!")
        else:
            val,top=pop(stack,top)
            print(f"Popped item: {val}")
    elif op==3:
        if top<0:
            print("Stack is empty!!")
        else:
            val=peek(stack,top)
            print(f"Val: {val}")
    elif op==4:
        break
    else:
        print("Invalid Operation!!")
```
